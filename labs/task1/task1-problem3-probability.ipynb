## Частина 1. Результати іспиту

Усі студенти належать до однієї з двох груп: $A$ — добре підготовлені до іспиту, $B$ — не дуже добре підготовлені. Нехай $E$ — подія «випадково вибраний студент складає іспит». Відомо:

$$P(B)=0.1,\qquad P(E\mid A)=0.8,\qquad P(E\mid B)=0.3.$$

- **(а)** Знайдіть імовірність того, що студент складе іспит **і** належить до групи $A$: $P(E\cap A)$.
- **(б)** Знайдіть імовірність скласти іспит: $P(E)$.
- **(в)** Знайдіть імовірність належати до групи $B$ за умови, що студент склав іспит: $P(B\mid E)$.

Наведіть формули та поясніть результати.


```python
# Відомі ймовірності
P_B = 0.1
P_E_given_A = 0.8
P_E_given_B = 0.3

# Оскільки студент належить або до A, або до B
P_A = 1 - P_B

# (a) Ймовірність одночасно належати до A і скласти іспит
P_E_and_A = P_E_given_A * P_A

# (б) Загальна ймовірність скласти іспит:
# формула повної ймовірності
P_E = P_E_given_A * P_A + P_E_given_B * P_B

# (в) Ймовірність належати до B за умови, що студент склав іспит:
# формула Байєса
P_B_given_E = (P_E_given_B * P_B) / P_E

print(f"P(A) = {P_A:.2f}")
print(f"P(E ∩ A) = {P_E_and_A:.2f}")
print(f"P(E) = {P_E:.2f}")
print(f"P(B | E) = {P_B_given_E:.2f}")
```

### Інтерпретація / виведення

Оскільки $P(B)=0.1$, маємо $P(A)=0.9$.

**(а)** Ймовірність того, що студент одночасно належить до групи $A$ і складе іспит:

$$
P(E\cap A)=P(E\mid A)P(A)=0.8\cdot0.9=0.72.
$$

Отже, **72%** усіх студентів належать до групи $A$ і складають іспит.

**(б)** За формулою повної ймовірності:

$$
P(E)=P(E\mid A)P(A)+P(E\mid B)P(B)
=0.8\cdot0.9+0.3\cdot0.1
=0.75.
$$

Отже, випадково вибраний студент складає іспит з ймовірністю **75%**.

**(в)** За формулою Байєса:

$$
P(B\mid E)
=
\frac{P(E\mid B)P(B)}{P(E)}
=
\frac{0.3\cdot0.1}{0.75}
=
0.04.
$$

Отже, серед студентів, які склали іспит, приблизно **4%** належать до групи $B$.

## Частина 2. Барбекю

На барбекю-вечірці є п'ять вегетаріанців. Кожна порція овочів підгорає вщент з імовірністю 0.6. Зазначте припущення про незалежність підгоряння різних порцій та обчисліть імовірності того, що:

- **(а)** жодна з п'яти порцій не підгорить;
- **(б)** рівно дві з п'яти порцій підгорять;
- **(в)** принаймні три з п'яти порцій підгорять.


### Розв'язання

Нехай $X$ — кількість підгорілих порцій. Оскільки є 5 порцій, кожна підгорає незалежно з ймовірністю $p=0.6$,

$$
X\sim Bin(5,0.6).
$$

Для біноміального розподілу:

$$
P(X=k)=\binom{n}{k}p^k(1-p)^{n-k}.
$$

**(а) Жодна порція не підгорить**

$$
P(X=0)=(1-0.6)^5=0.4^5=0.01024.
$$

Отже, ймовірність того, що жодна порція не підгорить, становить приблизно **1.02%**.

**(б) Рівно дві порції підгорять**

$$
P(X=2)=\binom{5}{2}(0.6)^2(0.4)^3.
$$

$$
P(X=2)=10\cdot0.36\cdot0.064=0.2304.
$$

Отже, ймовірність того, що рівно дві порції підгорять, становить **23.04%**.

**(в) Принаймні три порції підгорять**

$$
P(X\ge3)=P(X=3)+P(X=4)+P(X=5).
$$

$$
P(X\ge3)=0.68256.
$$

Отже, ймовірність того, що підгорять принаймні три з п'яти порцій, становить приблизно **68.26%**.

## Частина 3. Нормальний розподіл дохідності та портфель

Щотижнева дохідність акції $X$ має нормальний розподіл із середнім $\mu=0.03$ та **дисперсією** $\sigma^2=0.02$. Отже, стандартне відхилення дорівнює $\sqrt{0.02}$, а не $0.02$.

- **(а)** Обчисліть $P(X>0.01)$, $P(X\geq0.01)$, $P(X<0.01)$ та $P(0.01<X<0.03)$.
- **(б)** У 10 таких незалежних акцій інвестовано порівну 40% усього капіталу, а решту 60% — у безризиковий актив із дохідністю $r_f=0.01$. Визначте вид розподілу дохідності портфеля та обчисліть його параметри. Вага кожної акції становить $0.4/10$.
- **(в)** Поясніть вплив кореляції між активами на дисперсію інвестицій. Наведіть формулу дисперсії портфеля з коваріаційними доданками.


```python
# Параметри розподілу дохідності акції
stock_mean = 0.03
stock_variance = 0.02
stock_std = np.sqrt(stock_variance)

# Створюємо нормальний розподіл з заданими параметрами
stock_distribution = stats.norm(
    loc=stock_mean,
    scale=stock_std
)

# P(X > 0.01)
prob_gt_001 = 1 - stock_distribution.cdf(0.01)

# Для неперервного розподілу P(X >= 0.01) = P(X > 0.01),
# оскільки P(X = 0.01) = 0.
prob_ge_001 = prob_gt_001

# P(X < 0.01)
prob_lt_001 = stock_distribution.cdf(0.01)

# P(0.01 < X < 0.03)
prob_between = (
    stock_distribution.cdf(0.03)
    - stock_distribution.cdf(0.01)
)

print(f"P(X > 0.01) = {prob_gt_001:.4f}")
print(f"P(X >= 0.01) = {prob_ge_001:.4f}")
print(f"P(X < 0.01) = {prob_lt_001:.4f}")
print(f"P(0.01 < X < 0.03) = {prob_between:.4f}")
```

### Частина 3(б). Дохідність портфеля

Позначимо дохідності акцій через $X_1,\ldots,X_{10}$. Вага кожної акції дорівнює $0.4/10=0.04$, тому дохідність портфеля:

$$
R_p=0.04\sum_{i=1}^{10}X_i+0.6\cdot0.01.
$$

Сума незалежних нормально розподілених величин також має нормальний розподіл. Безризикова частина додає сталу дохідність, тому $R_p$ має нормальний розподіл.

Його середнє:

$$
E[R_p]=0.04\cdot10\cdot0.03+0.6\cdot0.01=0.018.
$$

Через незалежність акцій їхні коваріації дорівнюють нулю. Безризиковий актив має нульову дисперсію, тому

$$
\operatorname{Var}(R_p)=10\cdot(0.04)^2\cdot0.02=0.00032.
$$

Отже, $R_p\sim N(0.018,0.00032)$, де другий параметр — дисперсія. Очікувана щотижнева дохідність становить **1.8%**, а стандартне відхилення — $\sqrt{0.00032}\approx0.01789$, тобто **1.79 відсоткового пункту**.

### Частина 3(в). Вплив кореляції

Якщо дохідності активів залежать одна від одної, дисперсія портфеля обчислюється так:

$$
\operatorname{Var}(R_p)=\sum_i w_i^2\sigma_i^2
+2\sum_{i<j}w_iw_j\operatorname{Cov}(X_i,X_j),
\qquad
\operatorname{Cov}(X_i,X_j)=\rho_{ij}\sigma_i\sigma_j.
$$

Тут $w_i$ — вага активу, $\sigma_i$ — стандартне відхилення його дохідності, а $\rho_{ij}$ — кореляція дохідностей двох активів. Безризиковий актив не додає дисперсії чи коваріації.

За додатних ваг додатна кореляція збільшує дисперсію портфеля: ціни активів частіше рухаються в одному напрямку. Від’ємна кореляція зменшує її, бо падіння дохідності одного активу може частково компенсуватися зростанням іншого. Тому розподіл коштів між активами зменшує ризик сильніше, коли їхні дохідності мають слабку або від’ємну кореляцію. У пункті (б) акції незалежні, тому всі коваріаційні доданки дорівнюють нулю.


## Частина 4. Щотижнева дохідність Tesla

За допомогою Python (наприклад, `yfinance`) завантажте щотижневі ціни Tesla Inc. (`TSLA`) **за останні два роки** та обчисліть щотижневу дохідність. Зафіксуйте дату завантаження, точні межі дворічного періоду, вибір скоригованих або нескоригованих цін і спосіб обробки неповних тижнів. Збережіть копію даних у `data/` для відтворюваності.

- **(а)** За емпіричним розподілом оцініть імовірності дохідності **строго нижче** $-2\%$ та $-4\%$.
- **(б)** Оцініть ті самі ймовірності за нормальним розподілом із параметрами, що дорівнюють вибірковому середньому та вибірковій дисперсії дохідності.
- **(в)** Порівняйте результати й обговоріть переваги та недоліки обох методів.

Для емпіричних оцінок врахуйте строгість нерівності на порозі; дохідність обчислюйте як відносну зміну ціни між послідовними тижнями.


```python
import yfinance as yf

# Фіксуємо дату завантаження, щоб результати можна було відтворити.
download_date = pd.Timestamp("2026-10-07")

# Точний двохрічний період.
period_start = download_date - pd.DateOffset(years=2)

# yfinance трактує end як невключну межу,
# тому додаємо один день.
period_end_exclusive = download_date + pd.Timedelta(days=1)

# Файл для локальної копії даних.
tsla_cache_path = DATA_DIR / "tsla_daily_2024-10-07_2026-10-07.csv"

if tsla_cache_path.exists():
    # Якщо дані вже завантажені раніше — читаємо локальну копію.
    tsla_daily = pd.read_csv(
        tsla_cache_path,
        header=[0, 1],
        index_col=0,
        parse_dates=True
    )
else:
    # Завантажуємо денні ціни Tesla.
    tsla_daily = yf.download(
        "TSLA",
        start=period_start.strftime("%Y-%m-%d"),
        end=period_end_exclusive.strftime("%Y-%m-%d"),
        auto_adjust=False,
        progress=False
    )

    # Зберігаємо копію для відтворюваності.
    tsla_daily.to_csv(tsla_cache_path)

print(f"Дата завантаження: {download_date.date()}")
print(f"Період: {period_start.date()} — {download_date.date()}")
```

```python
# Беремо скориговану ціну закриття.
# Adj Close враховує корпоративні події, наприклад спліти.
if isinstance(tsla_daily.columns, pd.MultiIndex):
    tsla_adjusted_close = tsla_daily["Adj Close"]["TSLA"]
else:
    tsla_adjusted_close = tsla_daily["Adj Close"]

tsla_adjusted_close = tsla_adjusted_close.dropna()
```

```python
# Визначаємо останню повністю завершену п'ятницю.
# Поточний неповний тиждень не використовуємо.
days_back = (download_date.weekday() - 4) % 7

if days_back == 0:
    days_back = 7

last_completed_friday = download_date - pd.Timedelta(days=days_back)

# Для кожного тижня беремо останню доступну торгову ціну.
# Якщо п'ятниця була вихідним, буде взята остання ціна цього тижня.
tsla_weekly_price = (
    tsla_adjusted_close
    .resample("W-FRI")
    .last()
    .dropna()
)

# Видаляємо неповний поточний тиждень.
tsla_weekly_price = tsla_weekly_price[
    tsla_weekly_price.index <= last_completed_friday
]

# Щотижнева дохідність як відносна зміна ціни
tsla_weekly_return = tsla_weekly_price.pct_change().dropna()

print(f"Останній використаний завершений тиждень: {last_completed_friday.date()}")
print(f"Кількість тижневих дохідностей: {len(tsla_weekly_return)}")
```

```python
# Строго нижче -2% та -4%, як вимагає умова.
empirical_below_2 = np.mean(tsla_weekly_return < -0.02)
empirical_below_4 = np.mean(tsla_weekly_return < -0.04)

print(
    f"Емпірична P(R < -2%) = "
    f"{empirical_below_2:.4f} ({empirical_below_2 * 100:.2f}%)"
)

print(
    f"Емпірична P(R < -4%) = "
    f"{empirical_below_4:.4f} ({empirical_below_4 * 100:.2f}%)"
)
```

```python
# Вибіркові параметри тижневої дохідності
weekly_mean = tsla_weekly_return.mean()
weekly_variance = tsla_weekly_return.var(ddof=1)
weekly_std = np.sqrt(weekly_variance)

# Ймовірності за підігнаним нормальним розподілом
normal_below_2 = stats.norm.cdf(
    -0.02,
    loc=weekly_mean,
    scale=weekly_std
)

normal_below_4 = stats.norm.cdf(
    -0.04,
    loc=weekly_mean,
    scale=weekly_std
)

print(f"Вибіркове середнє = {weekly_mean:.4f}")
print(f"Вибіркова дисперсія = {weekly_variance:.6f}")
print(f"Вибіркове стандартне відхилення = {weekly_std:.4f}")

print(
    f"Нормальна P(R < -2%) = "
    f"{normal_below_2:.4f} ({normal_below_2 * 100:.2f}%)"
)

print(
    f"Нормальна P(R < -4%) = "
    f"{normal_below_4:.4f} ({normal_below_4 * 100:.2f}%)"
)
```

```python
probability_comparison = pd.DataFrame({
    "Емпірична": [
        empirical_below_2,
        empirical_below_4
    ],
    "Нормальна": [
        normal_below_2,
        normal_below_4
    ]
}, index=[
    "R < -2%",
    "R < -4%"
])

display((probability_comparison * 100).round(2).rename(
    columns={
        "Емпірична": "Емпірична, %",
        "Нормальна": "Нормальна, %"
    }
))
```

## Частина 5. Біноміальна модель ціни

Початкова ціна активу $S(0)=10$. У кожному періоді ціна зростає з імовірністю $0.55$, а дохідність при русі вгору та вниз дорівнює $0.02$ і $-0.01$ відповідно. Зазначте припущення про незалежність рухів ціни між періодами.

**(а)** Обчисліть імовірності того, що ціна зросте:

1. п'ять періодів поспіль;
2. три періоди поспіль;
3. рівно у трьох із п'яти періодів;
4. рівно у десяти з п'ятнадцяти періодів.

**(б)** Обчисліть очікувану сукупну дохідність інвестиції після двох періодів, враховуючи послідовне множення коефіцієнтів зростання ціни.


### Розв'язання

Нехай у кожному періоді ціна зростає з ймовірністю

$$
p=0.55,
$$

а падає з ймовірністю

$$
q=1-p=0.45.
$$

Припускаємо, що рухи ціни в різних періодах незалежні.

#### (а) Ймовірності зростання ціни

**1. Ціна зростає п'ять періодів поспіль**

Для цього зростання має відбутися у кожному з п'яти незалежних періодів:

$$
P=0.55^5\approx0.0503.
$$

Отже, ймовірність становить приблизно **5.03%**.

**2. Ціна зростає три періоди поспіль**

$$
P=0.55^3\approx0.1664.
$$

Отже, ймовірність становить приблизно **16.64%**.

**3. Ціна зростає рівно у трьох із п'яти періодів**

Кількість зростань має біноміальний розподіл:

$$
X\sim Bin(5,0.55).
$$

Тому

$$
P(X=3)
=
\binom{5}{3}(0.55)^3(0.45)^2
\approx0.3369.
$$

Отже, ймовірність становить приблизно **33.69%**.

**4. Ціна зростає рівно у десяти з п'ятнадцяти періодів**

$$
X\sim Bin(15,0.55),
$$

тому

$$
P(X=10)
=
\binom{15}{10}(0.55)^{10}(0.45)^5
\approx0.1404.
$$

Отже, ймовірність становить приблизно **14.04%**.

#### (б) Очікувана сукупна дохідність після двох періодів

При зростанні ціна множиться на

$$
1+0.02=1.02,
$$

а при падінні — на

$$
1-0.01=0.99.
$$

Очікуваний коефіцієнт зміни ціни за один період:

$$
E[M]
=
0.55\cdot1.02+0.45\cdot0.99
=
1.0065.
$$

Через незалежність двох періодів очікуваний коефіцієнт зміни за два періоди:

$$
E[M_1M_2]
=
E[M_1]E[M_2]
=
1.0065^2
=
1.01304225.
$$

Тому очікувана сукупна дохідність:

$$
1.01304225-1
=
0.01304225.
$$

Отже, очікувана дохідність інвестиції після двох періодів становить приблизно **1.30%**.

## Частина 6. Неперервні розподіли

Для кожного з розподілів $N(0,1)$, $t_2$ (розподіл Стьюдента з 2 ступенями свободи), $\operatorname{Exp}(1)$ (експоненціальний розподіл з інтенсивністю $\lambda=1$) та $\chi^2_3$ (розподіл хі-квадрат із 3 ступенями свободи) обчисліть:

- **(а)** $P(1<X<1.5)$;
- **(б)** $P(1\leq X<1.5)$;
- **(в)** $P(X>1)$;
- **(г)** $P(X\leq1.5)$;
- **(ґ)** $P(X=1.5)$.

Поясніть, як неперервність розподілів впливає на ймовірності в окремих точках та на включення меж інтервалу.


```python
# Розподіли, які задані в умові
continuous_distributions = {
    "N(0,1)": stats.norm(loc=0, scale=1),
    "t(2)": stats.t(df=2),

    # Для Exp(λ=1) параметр scale = 1/λ = 1
    "Exp(1)": stats.expon(scale=1),

    # Хі-квадрат із 3 ступенями свободи
    "χ²(3)": stats.chi2(df=3)
}

continuous_results = []

for name, distribution in continuous_distributions.items():
    # (а) P(1 < X < 1.5)
    prob_1_to_15 = (
        distribution.cdf(1.5)
        - distribution.cdf(1)
    )

    # (б) Для неперервного розподілу включення точки X=1
    # не змінює ймовірність
    prob_1_inclusive_to_15 = prob_1_to_15

    # (в) P(X > 1)
    # sf(x) = 1 - F(x), тобто ймовірність правого хвоста
    prob_above_1 = distribution.sf(1)

    # (г) P(X <= 1.5)
    prob_below_15 = distribution.cdf(1.5)

    # (ґ) Для неперервної випадкової величини
    # ймовірність одного конкретного значення дорівнює 0
    prob_equal_15 = 0.0

    continuous_results.append({
        "Розподіл": name,
        "P(1 < X < 1.5)": prob_1_to_15,
        "P(1 ≤ X < 1.5)": prob_1_inclusive_to_15,
        "P(X > 1)": prob_above_1,
        "P(X ≤ 1.5)": prob_below_15,
        "P(X = 1.5)": prob_equal_15
    })

continuous_results_df = pd.DataFrame(continuous_results)

display(
    continuous_results_df.set_index("Розподіл").round(4)
)
```

### Інтерпретація

Для всіх чотирьох розподілів виконується

$$
P(1<X<1.5)=P(1\le X<1.5),
$$

оскільки всі вони є неперервними, а для неперервної випадкової величини ймовірність отримати одне конкретне значення дорівнює нулю:

$$
P(X=1)=P(X=1.5)=0.
$$

Тому включення або невключення межі інтервалу не змінює його ймовірність.

Для стандартного нормального розподілу $N(0,1)$ маємо $P(X>1)=0.1587$, а для розподілу Стьюдента $t_2$ — $0.2113$. Більша ймовірність у $t_2$ пов'язана з тим, що цей розподіл має важчі хвости.

Для $Exp(1)$ маємо $P(X>1)=0.3679$, а для $\chi^2_3$ — $0.8013$. Ці розподіли мають лише невід'ємні значення і є правосторонньо асиметричними, тому їхні ймовірності в заданих інтервалах суттєво відрізняються від симетричних $N(0,1)$ та $t_2$.

Головний висновок: у неперервних розподілах ймовірність визначається площею на інтервалі, а ймовірність окремої точки завжди дорівнює нулю.
